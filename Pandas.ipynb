## GroupBy

```python
#SELECT region, SUM(sales) AS total_sales FROM Orders GROUP BY region;

orders.groupby('region').sales.sum()
```

```python
# SELECT region, SUM(sales), AVG(profit), COUNT(*) FROM Orders GROUP BY region;
(
    orders
    .groupby('region')
    .agg(
        total_sales = ('sales', 'sum'),
        avg_profit = ('profit', 'mean'),
        no_of_counts = ('order_id', 'count')
    )
)
```

```python
# SELECT region,AVG(profit)FROM Orders GROUP BY region;

orders.groupby('region').profit.mean()
```

```python
# SELECT category, AVG(profit), MAX(sales), MIN(sales) FROM Orders GROUP BY category;

orders.groupby('category').agg(
    avg_profit = ('profit', 'mean'),
    max_sales = ('sales', 'max'),
    min_sales =  ('sales', 'min')
)
```

# JOINS => merge


### 1. Inner Join

```python
df1 = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "Eve"],
    "Age": [25, 30, 22, 35]
})

df2 = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "Charlie"],
    "Salary": [50000, 55000, 40000, 48000]
})

#select * from df1 inner join df2 ON df1.name = df2.name

df1.merge( df2, on= 'Name', how= 'inner' )

```

### 2. Left Join

```python
df1.merge( df2, on= 'Name', how= 'left' )
```

### 3. right Join

```python
df1.merge( df2, on= 'Name', how= 'right' )
```

### 4. Full Join

```python
df1.merge( df2, on= 'Name', how= 'outer' )
```
